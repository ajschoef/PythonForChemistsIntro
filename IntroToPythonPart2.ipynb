## **Part 2B: Math with Numbers**
A lot of math in Python can look similar to the math you're used to:

* `2*x`
  *  The product of 2 and a variable (`x`)
* `y = 27 + z`
  *  The variable `y` is the sum of an integer (`27`) and another variable (`x`).
* `score = examscore + curve`
  * the sum of two variables

But there are some operations in Python that look decidedly non-algebraic. Here's an example that comes up frequently:

`count = 0`

`count = count + 1`

In this case, the first line of code instantiates (or creates) the variable `count`, and the second line *re*defines it as a sum with itself. After both lines of code are run, the value of `count` will be 1. This might seem silly or useless, but I can promise you that it's not. For now, you'll have to trust me.

## **Part 2E: Math with Arrays**

One of the most important benefits of conducting mathematical operations using code is the benefit of **scale**. Arrays help you access that scale quickly and easily. An array is like a list that can handle math. Run the examples below to get a sense for how this works.

```python
import numpy

scores = [78, 82, 69, 87, 91, 88]
scorearray = numpy.array(scores)

curve = 9

curvedscores = scorearray + curve
print(curvedscores)
```

Notice how differently this works compared to the failed code in the List Math example above. When working with an array, the math maps to each element in the array. Now look at the example below and compare these results to how this worked with the list example.

```python
import numpy
scores = [0.65, 0.75, 0.91, 0.59, 0.82, 0.88]
scorearray = numpy.array(scores)
percentages = scorearray*100
print(percentages)
```

When you're adding or subtracting arrays, it's important to remember that they must be the same size. Run the code block below and study the error message that results.

```python
import numpy
scores = [78, 82, 69, 87, 91, 88]
scorearray = numpy.array(scores)
bonus = [5, 3, 2, 5]
bonusarray = numpy.array(bonus)

adjustedscores = scorearray + bonusarray
print(adjustedscores)
```

Can you tell, based on the error message, what went wrong? (Hint: Look at the very end of the error message. That's usually the most useful part.)
