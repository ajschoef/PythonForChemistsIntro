## **Part 4A: Loops** ##
Loops are a way of running code repetitively over a series of items or elements. There are two types of loops: `for` loops and `while` loops. We'll cover both in this section.

### **`for` loops** ###
One of the simplest ways to set up a loop iteration is to run it through a list. Have a look at the example below; run it to see what happens.

```python
lunch = ['soup', 'salad', 'sandwich']

for plate in lunch:
  print(plate)
```

Look carefully at the `for` line in the code above and notice the following aspects:



1.   The `for` line is the loop initiation line.
2.   In the initiation line, the loop is instantiated with the `for` command
3.   The second part of the loop initiation line is the loop variable. In this case, the loop variable is `plate`. You can give a loop variable any name you wish--common loop variables are `i` and `n`. The loop variable is a placeholder variable, and its value changes with each iteration through the loop.
4.   The term `in` in this line points the loop at the data structure for iteration.
5.   `lunch` is the data structure we're iterating over. In this case, it's a list of three strings, but you can iterate over a variety of data structures.
6.  The loop initiation line ends with a colon: `:`.

After the loop initiation line comes the body of the loop. These are the commands that will be executed with each pass through the loop, and all of them are indented by one tab. In this case, it's just a print statement. Notice how the string that is printed changes each time as the loop runs through each element of the loop.



The `for` loop above simply printed list elements. The example below is different--it performs a mathematical operation with every pass. Run the code and notice how the value of `count` changes at the end of the loop.

```python
count = 0
print(count)

years = [2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026]

for year in years:
  count = count + 1

print(count)
```

There are three important points to understand about the loop above.

 First of all, notice that the variable `count` had to be instantiated outside the loop with a value of 0.

 Secondly, have a look at the last line of code (`print(count)`). This line is **not** tabbed in. That means it executes only after the loop is completely finished.  

 Finally, and least obviously, notice that it doesn't matter what the elements are in `years`. It only matters how many there are. Consider the example below:

```python
count = 0
print(count)

random = ['scarf', 'quokka', 'sandy', 'Enceladus', 'lithium', 'cuddly', 'green', 'pointy']

for stuff in random:
  count = count + 1

print(count)
```

See how the output is exactly the same? There are eight elements in `random` just like there were eight elements in `years`.

### **`while` loops** ###

`for` loops iterate over every element in a data structure. `while` loops, on the other hand, keep executing until a condition fails. Have a look at the example below; be sure to run the code so you can see what happens.

```python
x = 0
while x < 10:
  print(x)
  print("still going...")
  x = x + 1
```

In the example above, what would happen if the `x = x + 1` line was omitted? Think about it for a moment, but **don't** actually test your theory!

The answer: It would create an infinite loop. The condition (x < 10) would always hold true because x would never change.

### Pro Tip! ###
If you create an infinite loop that won't stop running, you can type Ctrl-C to interrupt it.

## **Part 4B: Conditionals** ##

Conditionals are sets of code that execute in response to conditions you set. In this way, a conditional is a close relative to the `while` loop. There are three key commands that form part of a conditional: `if`, `elif`, and `else`.

The code block below shows you an example of a simple conditional. Try changing the value of `value` or `sweetspot` to see what happens.

```python
sweetspot = 5

value = 4

if value < sweetspot:
  print ('not enough')
elif value > sweetspot:
  print('too much')
elif value == sweetspot:
  print('just right')
```

Notice that the first conditional is housed in an `if` statement, and the subsequent conditionals are housed in `elif` statements ("elif" stands for "else if"). There's no `else` statement in the example above, but `else` statements are easy enough to understand. An `else` statement is a catch-all for anything that doesn't fit into any of the other conditionals. Have a look at the example below, then change the values of `limit` and `value` to see what happens.



```python
limit = 10
value = 5

if value <= limit:
  print('okay')
else:
  print('nope')

```
