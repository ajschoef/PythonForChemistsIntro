### **Part 3C: Plotting Multiple Data Sets** ###

It's simple to plot multiple data sets. You just need to use multiple `matpotlib.pyplot.plot` functions followed by a single `matplotlib.pyplot.show` function at the end.

In the code block below, there are data for three lines, followed by three plotting functions and a single show function. Run the code to see what happens, then change the slopes and intercepts for the three data sets to see how the plot is affected.

```python
import numpy
import matplotlib

# below is the x array, which will be used for all three data sets
x = numpy.linspace(1,50,10)

# below, the slope and intercept values are defined for the three different lines
m1 = 2
yint1 = 1

m2 = -1
yint2 = -2

m3 = 5
yint3 = 4

# the code below calculates three different y sets based on the slopes and intercepts above.
y1 = m1*x + yint1
y2 = m2*x + yint2
y3 = m3*x + yint3

# The code below sets up the datasets for plotting and then displays the plot
matplotlib.pyplot.plot(x, y1)
matplotlib.pyplot.plot(x, y2)
matplotlib.pyplot.plot(x, y3)
matplotlib.pyplot.show()
```

Notice in the plot above that MatPlotLib automatically changes the color of the lines to make it easier to distinguish them. You can do this yourself, too, using the `color` function. Consider the example below.

```python
import numpy
import matplotlib

x = numpy.linspace(1,50,10)

m1 = 2
yint1 = 1
m2 = -1
yint2 = 30

y1 = m1*x + yint1
y2 = m2*x + yint2

# The color function is embedded in the plotting function below.
matplotlib.pyplot.plot(x, y1, color = 'maroon')
matplotlib.pyplot.plot(x, y2, color = 'gold')
matplotlib.pyplot.show()
```

There are many options for selecting colors in MatPlotLib. In the code above, you can try out different color names to see if they work, just for fun.
