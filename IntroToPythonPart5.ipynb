# **Custom Functions** #

You already know what a function is. It's a repeatable command. Here are some functions you've already learned:


*   `print()`
*   `matplotlib.pyplot.plot()`
*   `numpy.linspace()`

These are functions that have already been built, but you can also create your own custom function. Here's an example; be sure to run the code so you can see the results:



```python
def myfunction(x, y):
  z = 2 * x + y
  return z
```

Let's look at the code above line-by-line and piece-by-piece.



*   `def` is the command to start a function definition.
*   `myfunction` is the name of the custom function. You can make this anything you want, as long as you're adhering to variable naming rules and not using an existing command or function name.
*   The parentheses after the function name house the arguments. Arguments are variables or data elements that you give to the function, and these are the only things it can receive from the "outside world" of the code. In this case, the arguments are `x` and `y`.
*   The function `def` line ends with a colon (`:`).

After the function definition line comes the body of the function. These are the commands that are executed each time the function runs. Notice that they are all tabbed-in. In this case, there is a mathematical equation defining a value for the variable `z` based on the arguments `x` and `y`.

The very last line in the body of the function is a `return` line. If you want to pass a value, a variable, or a data structure back out of the function and into the main code, you must do it through the return function. In this case, the function returns `z`.

If you run the code above (go ahed and try it), nothing will happen. That's because while the function has been **defined**, it hasn't been **executed**. You've built it, but you haven't used it. The code block below both **defines** and **excutes** the function. Run it and see what happens.

```python
def myfunction(x, y):
  z = 2 * x + y
  return z

#trial 1
trial1 = myfunction(1,1)
print(trial1)

#trial 2
trial2 = myfunction(2,-7)
print(trial2)

#trial 3
number1 = 20
number2 = -2
trial3  = myfunction(number1, number2)
print(trial3)
```

In the example above, the function is run three different times, and the results are stored in `trial1`, `trial2`, and `trial3`.

Notice that for each trial, the arguments are different. In trial 1, 1 is passed to `x` and 1 is passed to `y`. In trial 2, 2 is passed to `x` and -7 is passed to `y`. In trial 3, variables (`number1` and `number2`) are passed to `x` and `y`. (Note that the variable had to be defined outside the function first.) There's an incredibly important concept at work, here: **the locality of variables in python functions**. To get a sense for what this means, study the code block below and then run it.
